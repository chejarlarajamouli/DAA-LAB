```python

edges = [
    (1, 'A', 'B'),
    (2, 'A', 'C'),
    (3, 'C', 'D'),
    (4, 'B', 'C'),
    (5, 'B', 'D')
]

# Sort edges by weight
edges.sort()

parent = {}

# Make each vertex its own parent
for w, u, v in edges:
    parent[u] = u
    parent[v] = v

# Find parent
def find(x):
    while parent[x] != x:
        x = parent[x]
    return x

print("Minimum Spanning Tree:")

cost = 0

for w, u, v in edges:

    x = find(u)
    y = find(v)

    if x != y:
        print(u, "-", v, "=", w)
        cost += w
        parent[y] = x

print("Total Cost =", cost)
```
