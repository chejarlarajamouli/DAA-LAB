```python
# Floyd-Warshall Algorithm

INF = float('inf')

# Graph represented as an adjacency matrix
# INF means there is no direct edge between vertices

graph = [
    [0,   5,   INF, 10],
    [INF, 0,   3,   INF],
    [INF, INF, 0,   1],
    [INF, INF, INF, 0]
]

V = len(graph)

# Initialize distance matrix
dist = [row[:] for row in graph]

# Floyd-Warshall algorithm
for k in range(V):
    for i in range(V):
        for j in range(V):
            if dist[i][k] != INF and dist[k][j] != INF:
                dist[i][j] = min(
                    dist[i][j],
                    dist[i][k] + dist[k][j]
                )

# Print shortest distance matrix
print("All-Pairs Shortest Distance Matrix:")
for i in range(V):
    for j in range(V):
        if dist[i][j] == INF:
            print("INF", end="\t")
        else:
            print(dist[i][j], end="\t")
    print()
```
