```python
# Travelling Salesman Problem (TSP)
# Nearest Neighbour Algorithm
# Starting from A, B, C, and D

cities = ['A', 'B', 'C', 'D']

distance = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

def tsp(start):
    n = len(cities)
    visited = [False] * n
    route = [start]
    visited[start] = True
    total_distance = 0
    current = start

    for _ in range(n - 1):
        nearest = -1
        min_distance = float('inf')

        for j in range(n):
            if not visited[j] and distance[current][j] < min_distance:
                min_distance = distance[current][j]
                nearest = j

        route.append(nearest)
        visited[nearest] = True
        total_distance += min_distance
        current = nearest

    # Return to the starting city
    total_distance += distance[current][start]
    route.append(start)

    print("Starting City:", cities[start])
    print("Route:", " -> ".join(cities[i] for i in route))
    print("Total Distance:", total_distance)
    print("-" * 40)


# Run TSP from every city
for start in range(len(cities)):
    tsp(start)

```
